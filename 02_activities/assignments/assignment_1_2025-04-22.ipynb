```python
import scipy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
```

```python
# generating fake data
np.random.seed(613)
x= np.arange(50)
y= np.random.randint(0,100,50)

# creating scatterplot
plt.scatter(x,y)
```

```python

# creating line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
# creating histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
plt.show()
```

```python
# adding titles
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Growth')
ax.set_xlabel('Years')
plt.show()
```

```python
# defining font styles
font1 = {'family':'sans serif', 
         'color': 'black', 
         'size':20}
font2 = {'family':'monospace', 
         'color': 'blue', 
         'size':12}

# adding font styling to labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Growth', fontdict=font2)
ax.set_xlabel('Years', fontdict=font2)


```

```python
# changing marker styles
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y, marker='*', color='red')
```

```python
# changing line styles and widths
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y, marker='*', color='red', 
        linestyle='--', linewidth = 2)
```

```python
# changing edge styles
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y, marker='*', color='red', 
        linestyle='--', linewidth = 2, 
        markeredgecolor='orange')
```

### Activity
The following is my attempt at recreating the 2D histogram plot from the Python Graph Gallery. 

```python
# generating data
size = 100000

np.random.seed(613)
df = pd.DataFrame({
   'x': np.random.normal(size=size),
   'y': np.random.normal(size=size)
})
df.head()
```

```python
# recreating just the top right graph
fig, ax = plt.subplots(figsize=(8,8))

ax.hist2d(df['x'], df['y'], bins=(50, 50), cmap=plt.cm.jet)
ax.set_title('bins = (50, 50)')
plt.show()
```

**Notes:**
- aesthetic: the graph may benefit from x- and y-axis labels. So far there aren't any as this is just fake data.
- substantive: both x and y are random numbers generated from the same normal distribution and so are expected to be symmetric and distributionally identical to each other. This is accurately reflected in the graph gvien the circular ring. 
- perceptual: assuming that "red" signifies high frequency and "blue" signifies low frequency, then the graph clearly shows what values of x and y ahve high vs. low frequency. However, it would have been more clear if there was an additional legend to indicate this.
